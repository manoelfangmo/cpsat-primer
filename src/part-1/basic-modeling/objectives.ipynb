```python
#  Plain English: Maximizing 30 * x + 50 * y with constraints  x+ 10 * y <= 100 and x must be between -100 and 100 and Y must be between -100 and 100
# Math Version
# Maximize 30x + 50y
# subject to: x + 10y <= 100
#             -100 <= x <= 100, -100 <= y <= 100
#             x, y integers
from ortools.sat.python import cp_model

# Basic model with variables and constraints
model = cp_model.CpModel()
x = model.new_int_var(-100, 100, "x")
y = model.new_int_var(-100, 100, "y")
model.add(x + 10 * y <= 100)

# Maximize 30x + 50y
model.maximize(30 * x + 50 * y)

# Solve
solver = cp_model.CpSolver()
status = solver.solve(model)

# Print results
print(solver.status_name(status))
if status in (cp_model.OPTIMAL, cp_model.FEASIBLE):
    print(f"x = {solver.value(x)}")
    print(f"y = {solver.value(y)}")
    print(f"objective = {solver.objective_value}") # Value after x and y are plugged in
```

```python

# Plain English: 10 on/off switches, x0–x9.
# Even switches cost their number when ON; odd switches cost their number when OFF.
# Find the setting with the lowest total cost.
#
# Math: minimize  sum over even i of i * x_i  +  sum over odd i of i * (1 - x_i)
#       x_i ∈ {0, 1}
#
# Expected result: evens off, odds on, cost 0.model = cp_model.CpModel()
x_vars = [model.new_bool_var(f"x{i}") for i in range(10)]
expression = sum(i * x_vars[i] if i % 2 == 0 else i * ~x_vars[i] for i in range(10))
print(expression)
model.minimize(expression) 

solver = cp_model.CpSolver()
status = solver.solve(model)

print(solver.status_name(status))
if status in (cp_model.OPTIMAL, cp_model.FEASIBLE):
    for i, v in enumerate(x_vars):
        print(f"x{i} = {solver.value(v)}")
    print(f"objective = {solver.objective_value}")
```

```python
# Lexicographic Optimization; Two Or More Objective Function

# some basic model
model = cp_model.CpModel()
x = model.new_int_var(-100, 100, "x")
y = model.new_int_var(-100, 100, "y")
z = model.new_int_var(-100, 100, "z")
model.add(x + 10 * y - 2 * z <= 100)

# Define the objectives
first_objective = 30 * x + 50 * y
second_objective = 10 * x + 20 * y + 30 * z

# Optimize for the first objective
model.maximize(first_objective)
solver = cp_model.CpSolver()
solver.solve(model)

# Fix the first objective and optimize for the second
model.add(first_objective == int(solver.objective_value))  # fix previous objective: Ensures  fir objective is included  when solving for second objective
model.minimize(second_objective)  # optimize for second objective
status = solver.solve(model)

print(solver.status_name(status))
if status in (cp_model.OPTIMAL, cp_model.FEASIBLE):
    print(f"x = {solver.value(x)}")
    print(f"y = {solver.value(y)}")
    print(f"z = {solver.value(z)}")
    print(f"objective = {solver.objective_value}") # Value after x and y are plugged in
```

```python
## Absolute Value: Minimize the absolute value of 
# Assuming x is already defined in your model

model = cp_model.CpModel()
x = model.new_int_var(101, 200, "x")
abs_x = model.new_int_var(
    0, 100, "|x|"
)  # Create a variable to represent the absolute value of x
model.add_abs_equality(target=abs_x, expr=x)  # Define abs_x as the absolute value of x
model.minimize(abs_x)  # Set the objective to minimize abs_x
solver = cp_model.CpSolver()
status = solver.solve(model)

print(solver.status_name(status))
if status in (cp_model.OPTIMAL, cp_model.FEASIBLE):
    print(f"x = {solver.value(x)}")
    print(f"objective = {solver.objective_value}") # Value after x and y are plugged in
```
